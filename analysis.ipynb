## 6. Insights

### Insight 1 — Transaction Count vs Total Sales

Fashion accessories had the highest number of transactions,
but Food and beverages generated the highest total sales.

This shows that transaction count alone is not enough
to evaluate sales performance.

### Insight 2 — Branch Performance

Branch C had the highest total sales and the highest
average transaction value among the three branches.

### Insight 3 — Branch and Product Line Performance

The strongest-performing branch varied across product lines.
Branch C performed best in three product lines, Branch B in two,
and Branch A in one.
